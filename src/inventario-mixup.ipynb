# 1. Creación de tablas

Ejecuta las celdas de esta sección antes de cargar los datos.


### 1.1 Tablas padre


```sql
/*CREACION DE TABLAS*/

/*TABLAS PADRE*/
CREATE TABLE producto(
    idProducto NUMERIC(5,0) NOT NULL,
	nombre VARCHAR(50) NOT NULL,
	clasificacion VARCHAR(50) NOT NULL,
	precio NUMERIC(7,2) NOT NULL,
	precioDescuento NUMERIC(7,2) NULL,
	portada VARCHAR(50) NOT NULL,
	origen VARCHAR(50) NOT NULL,
	anio NUMERIC(4,0) NOT NULL,
	descripcion VARCHAR(100) NOT NULL,

	/*PRIMARY KEY*/
    CONSTRAINT pkProducto
	PRIMARY KEY (idProducto)
);
```

```sql
CREATE TABLE dia(
    idDia NUMERIC(5,0) NOT NULL,
	dia DATE NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkDia
	PRIMARY KEY (idDia)
);
```

```sql
CREATE TABLE responsable(
    idResponsable NUMERIC(5,0) NOT NULL,
    nombre VARCHAR(50) NOT NULL,
    paterno VARCHAR(50) NOT NULL,
    materno VARCHAR(50) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkResponsable
	PRIMARY KEY (idResponsable)
);
```

```sql
CREATE TABLE contacto(
    idContacto NUMERIC(5,0) NOT NULL,
    telefonoPrincipal CHAR(10) NOT NULL,
    telefonoEmergencia CHAR(10) NOT NULL,
    telefonoPrivado CHAR(10) NOT NULL,
    telefonoDistribuidor CHAR(10) NOT NULL,
    correoElectronico VARCHAR(50) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkContacto
	PRIMARY KEY (idContacto),

	/*UNIQUE*/
	CONSTRAINT uqContactoTelefonoPrincipal
	UNIQUE(telefonoPrincipal),

	CONSTRAINT uqContactoTelefonoEmergencia
	UNIQUE(telefonoEmergencia),

	CONSTRAINT uqContactoTelefonoPrivado
	UNIQUE(telefonoPrivado),

	CONSTRAINT uqContactoTelefonoDistribuidor
	UNIQUE(telefonoDistribuidor),

	CONSTRAINT uqContactoCorreoElectronico
	UNIQUE(correoElectronico)
);
```

```sql
CREATE TABLE codigopostal(
    idCodigoPostal NUMERIC(5,0) NOT NULL,
    codigoPostal CHAR(5) NOT NULL,
	alcaldia VARCHAR(50) NOT NULL,
    delegacion VARCHAR(50) NOT NULL,
    
	/*PRIMARY KEY*/
	CONSTRAINT pkCodigopostal
	PRIMARY KEY (idCodigoPostal)
);
```

```sql
CREATE TABLE idioma(
    idIdioma NUMERIC(5,0) NOT NULL,
    idioma VARCHAR(20) NOT NULL,
    tipo VARCHAR(20) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkIdioma
	PRIMARY KEY (idIdioma)
);
```

### 1.2 Tablas padre-hija


```sql
/*TABLAS PADRE-HIJA*/
CREATE TABLE cd(
    idProducto NUMERIC(5,0) NOT NULL,
	contenido VARCHAR(2) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkCd
	PRIMARY KEY (idProducto),

	/*FOREIGN KEY*/
	CONSTRAINT fkCdProducto
	FOREIGN KEY (idProducto)
	REFERENCES producto(idProducto)
);
```

```sql
CREATE TABLE sucursal(
    idSucursal NUMERIC(5,0) NOT NULL,
    nombre VARCHAR(50) NOT NULL,
    calle VARCHAR(50) NOT NULL,
    numero VARCHAR(10) NOT NULL,
    horarioInicio TIME NOT NULL,
    horarioFin TIME NOT NULL,
    piso VARCHAR(3) NULL,
    latitud CHAR(13) NOT NULL,
    longitud CHAR(13) NOT NULL,
    altitud VARCHAR(9) NOT NULL,
    idResponsable NUMERIC(5) NOT NULL,
    idContacto NUMERIC(5) NOT NULL,
    idCodigoPostal NUMERIC(5) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkSucursal
	PRIMARY KEY (idSucursal),

	/*FOREIGN KEY*/
	CONSTRAINT fkSucursalResponsable
	FOREIGN KEY (idResponsable)
	REFERENCES responsable(idResponsable),

    CONSTRAINT fkSucursalContacto
	FOREIGN KEY (idContacto)
	REFERENCES contacto(idContacto),

    CONSTRAINT fkSucursalCodigoPostal
	FOREIGN KEY (idCodigoPostal)
	REFERENCES codigopostal(idCodigoPostal),

	/*UNIQUE*/
	CONSTRAINT uqSucursalNombre
	UNIQUE(nombre),

	/*Ubicacion*/
	CONSTRAINT uqSucursalLatitudLongitudAltitud
	UNIQUE(latitud,longitud,altitud),

	/*Cardinalidad 1:1 con Responsable*/
	CONSTRAINT uqSucursalIdResponsable
	UNIQUE(idResponsable),

	/*Cardinalidad 1:1 con Contacto*/
	CONSTRAINT uqSucursalIdContacto
	UNIQUE(idContacto)

);
```

### 1.3 Tablas hijas y tablas puente


```sql
/*TABLAS HIJAS*/
CREATE TABLE dvd(
    idProducto NUMERIC(5,0) NOT NULL,
	tituloOriginal VARCHAR(50) NULL,
	contenido VARCHAR(2) NOT NULL,
	formato VARCHAR(50) NOT NULL,
	nombre VARCHAR(50) NOT NULL,
	paterno VARCHAR(50) NOT NULL,
	materno VARCHAR(50) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkDvd
	PRIMARY KEY (idProducto),

	/*FOREIGN KEY*/
	CONSTRAINT fkDvdProducto
	FOREIGN KEY (idProducto)
	REFERENCES producto(idProducto)
);
```

```sql
CREATE TABLE bluray(
    idProducto NUMERIC(5,0) NOT NULL,
    tituloOriginal VARCHAR(50) NULL,
    contenido VARCHAR(2) NOT NULL,
    nombre VARCHAR(50) NOT NULL,
    paterno VARCHAR(50) NOT NULL,
    materno VARCHAR(50) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkBluray
	PRIMARY KEY (idProducto),

	/*FOREIGN KEY*/
	CONSTRAINT fkBlurayProducto
	FOREIGN KEY (idProducto)
	REFERENCES producto(idProducto)
);
```

```sql
CREATE TABLE juego(
    idProducto NUMERIC(5,0) NOT NULL,
    contenido VARCHAR(2) NOT NULL,
    numeroJugadores NUMERIC(3,0) NOT NULL,
    edad NUMERIC(2,0) NOT NULL,
    genero VARCHAR(50) NOT NULL,
    marca VARCHAR(50) NOT NULL,
    nombre VARCHAR(50) NOT NULL,
    paterno VARCHAR(50) NOT NULL,
    materno VARCHAR(50) NOT NULL,
    formato VARCHAR(50) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkJuego
	PRIMARY KEY (idProducto),

	/*FOREIGN KEY*/
	CONSTRAINT fkJuegoProducto
	FOREIGN KEY (idProducto)
	REFERENCES producto(idProducto)
);
```

```sql
CREATE TABLE tracklist(
    idTracklist NUMERIC(5,0) NOT NULL,
    numero VARCHAR(3) NOT NULL,
    nombre VARCHAR(50) NOT NULL,
    idProducto NUMERIC(13,0) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkTracklist
	PRIMARY KEY (idTracklist),

	/*FOREIGN KEY*/
	CONSTRAINT fkTracklistCd
	FOREIGN KEY (idProducto)
	REFERENCES producto(idProducto)
);
```

```sql
CREATE TABLE productoxsucursal(
    idProducto NUMERIC(5,0) NOT NULL,
	idSucursal NUMERIC(5,0) NOT NULL,
    inventario NUMERIC(6,0) NOT NULL,
    stock NUMERIC(6,0) NOT NULL,
    venta NUMERIC(6,0) NOT NULL,
    oferta  NUMERIC(3) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkProductoxsucursal
	PRIMARY KEY (idSucursal,idProducto),

	/*FOREIGN KEY*/
	CONSTRAINT fkProductoxsucursalSucursal
	FOREIGN KEY (idSucursal)
	REFERENCES sucursal(idSucursal),

    CONSTRAINT fkProductoxsucursalProducto
	FOREIGN KEY (idProducto)
	REFERENCES producto(idProducto)
);
```

```sql
CREATE TABLE sucursalxdia(
    idSucursal NUMERIC(5,0) NOT NULL,
    idDia NUMERIC(5,0) NOT NULL,

	/*PRIMARY KEY*/
	CONSTRAINT pkSucursalxdia
	PRIMARY KEY (idSucursal,idDia),

	/*FOREIGN KEY*/
	CONSTRAINT fkSucursalxdiaSucursal
	FOREIGN KEY (idSucursal)
	REFERENCES sucursal(idSucursal),

    CONSTRAINT fkSucursalxdiaDia
	FOREIGN KEY (idDia)
	REFERENCES dia(idDia)
);
```

## Validación rápida

Esta consulta permite comprobar cuántas filas fueron cargadas en cada tabla principal.


```sql
SELECT 'PRODUCTO' AS tabla, COUNT(*) AS filas FROM producto
UNION ALL SELECT 'DIA', COUNT(*) FROM dia
UNION ALL SELECT 'RESPONSABLE', COUNT(*) FROM responsable
UNION ALL SELECT 'CONTACTO', COUNT(*) FROM contacto
UNION ALL SELECT 'CODIGOPOSTAL', COUNT(*) FROM codigopostal
UNION ALL SELECT 'IDIOMA', COUNT(*) FROM idioma
UNION ALL SELECT 'CD', COUNT(*) FROM cd
UNION ALL SELECT 'SUCURSAL', COUNT(*) FROM sucursal
UNION ALL SELECT 'DVD', COUNT(*) FROM dvd
UNION ALL SELECT 'BLURAY', COUNT(*) FROM bluray
UNION ALL SELECT 'JUEGO', COUNT(*) FROM juego
UNION ALL SELECT 'TRACKLIST', COUNT(*) FROM tracklist
UNION ALL SELECT 'PRODUCTOXSUCURSAL', COUNT(*) FROM productoxsucursal
UNION ALL SELECT 'SUCURSALXDIA', COUNT(*) FROM sucursalxdia
ORDER BY tabla;
```

## Validación de integridad

Comprobación adicional de la relación entre `bluray` y `producto`.


```sql
-- Validación específica de productos Blu-ray
SELECT
    b.idProducto,
    p.nombre
FROM bluray AS b
LEFT JOIN producto AS p
    ON b.idProducto = p.idProducto
ORDER BY b.idProducto;
```
